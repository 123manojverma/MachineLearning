## Isolation Forest Anomaly Detection

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Cluster 1
cluster1 = np.random.normal(
    loc=[-3, 5],
    scale=[0.4, 0.4],
    size=(100, 2)
)

# Cluster 2
cluster2 = np.random.normal(
    loc=[1, 1.5],
    scale=[0.6, 0.6],
    size=(100, 2)
)

# Anomalies
outliers = np.random.uniform(
    low=[-8, -6],
    high=[8, 10],
    size=(40, 2)
)

X = np.vstack([cluster1, cluster2, outliers])

plt.scatter(X[:,0], X[:,1])
plt.show()
```

```python
from sklearn.ensemble import IsolationForest
clf=IsolationForest(contamination=0.2)
clf.fit(X)
predictions=clf.predict(X)
```

```python
predictions
```

```python
abn_ind=np.where(predictions<0)
abn_ind
```

```python
plt.scatter(X[:,0],X[:,1])
plt.scatter(X[abn_ind,0],X[abn_ind,1],edgecolors="r")
```
