```python
corpus="""Artificial intelligence is changing the way we work and communicate.
Machine learning model's can identify patterns in large amounts of data.
"""
```

```python
corpus
```

```python
import nltk

nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
# Tokenization
# Sentence --> Paragraphs
from nltk.tokenize import sent_tokenize

document=sent_tokenize(corpus)
```

```python
type(document)
```

```python
for sentence in document:
    print(sentence)
```

```python
# Tokenization 
# Paragraph-->words
# Sentence-->words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in document:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
# Tokenizes text into words while handling punctuation and contractions

# Notice that it intelligently separates contractions ("I'm" → ['I', "'m"]) and punctuation into separate tokens.

from nltk.tokenize import TreebankWordTokenizer
tokenizer=TreebankWordTokenizer()
tokenizer.tokenize(corpus)
```
