## Wordnet Lemmatizer

Lemmatization technique is like stemming. The output we will get after lemmatization is called 'lemma', which is a root word rather than root stem, theo output of stemming. After lemmatization, we will be getting a valid word that means the same thing.

NLTK provides WordNetLemmatizer class which is a thin wrapper aroung the wordnet corpus. This class uses morphy() function to the WordNet CorpusRender class to find a lemma. Let us understand it with ane example -

```python
import nltk
nltk.download("wordnet")
nltk.download("omw-1.4")  # Optional but recommended
```

```python
# Q&A, Chatbots, Text summarization

from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
```

```python
'''
POS - Part of Speech
Noun - n
Verb - v
Adjective - a
Adverb -r 
'''
lemmatizer.lemmatize("going",pos='v')
```

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]

for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos='v'))
```

```python
lemmatizer.lemmatize("goes",pos='n')
```

```python
lemmatizer.lemmatize("fairly",pos='v'),lemmatizer.lemmatize("sprotingly",pos='v')
```
