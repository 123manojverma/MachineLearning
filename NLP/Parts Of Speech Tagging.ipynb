## Parts Of Speech Tags

```python
# CC cordinating conjunction
# CD cardinal digit
# DT deteminer
# EX existential there (like: "there is" ... think of it like "there exists")
# FW foreign word
# IN preposition/subordinating conjunction
# JJ adjective - 'big'
# JJR adjective, comparative - 'bigger'
# JJS adjective, superlative - 'biggest'
# LS list marker 1)
# MD modal - could,will
# NN noun,singular - 'desk'
# NNS noun,plural - 'desks'
# NNP  proper noun, singular - 'Harrison'
# NNPS proper noun, plural - 'Americans'
# PDT  predeterminer - 'all the kids'
# POS  possessive ending parent's
# PRP  personal pronoun - I, he, she
# PRP$ possessive pronoun - my, his, hers
# RB   adverb - very, silently,
# RBR  adverb, comparative - better
# RBS  adverb, superlative - best
# RP   particle - give up
# TO   - to go 'to' the store.
# UH   interjection - errrrrrrrm
# VB   verb, base form - take
# VBD  verb, past tense - took
# VBG  verb, gerund/present participle - taking
# VBN  verb, past participle - taken
# VBP  verb, sing. present, non-3d - take
# VBZ verb, 3rd person sing. present - takes
# WDT wh-determiner - which
# WP wh-pronoun - who, want
# WP$ possessive wh-pronoun, eg- whose
# WRB wh-adverb, eg- where,when
```

```python
paragraph="""Early on a bright Monday morning, Harrison walked through the crowded streets of the old city carrying a heavy backpack filled with books, maps, and a small notebook that his parents had given him years ago. He believed that every journey, no matter how short or ordinary it seemed, could teach something valuable if people paid careful attention to the details around them. As he crossed the marketplace, Americans, tourists, local artists, musicians, and excited children were all busy talking, laughing, bargaining, and taking photographs of colorful buildings and ancient monuments. Suddenly, someone shouted, "Hey, watch out!" and everyone quickly stepped aside as a cyclist rushed past at incredible speed. Harrison smiled, quietly apologized to an elderly woman whose basket had almost fallen, and then continued toward the library, where he planned to spend the afternoon reading about science, technology, history, and philosophy. Although the weather had been pleasant earlier, dark clouds gradually gathered across the sky, and within minutes a strong wind began blowing through the narrow streets. "Wow!" exclaimed a young boy as the first drops of rain landed on the pavement. Harrison carefully packed his notebook into a waterproof bag, took shelter beneath a large stone archway, and patiently waited until the storm became lighter. During the delay, he started an interesting conversation with a teacher who explained how curiosity, discipline, and persistence often mattered more than natural talent when solving difficult problems. Inspired by the discussion, Harrison promised himself that he would study more consistently, help others whenever possible, and never stop asking thoughtful questions, because knowledge grows not only from reading books but also from observing people, sharing experiences, and learning from every unexpected situation encountered along the way.
"""
```

```python
import nltk
sentences=nltk.sent_tokenize(paragraph)
```

```python
sentences
```

```python
# We wil find the Pos Tag
from nltk.corpus import stopwords
nltk.download('averaged_perceptron_tagger_eng')

for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[word for word in words if word not in set(stopwords.words('english'))]
    pos_tag=nltk.pos_tag(words)
    print(pos_tag)
```
